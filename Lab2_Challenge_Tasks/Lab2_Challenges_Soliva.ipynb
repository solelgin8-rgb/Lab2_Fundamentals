```python
# [EXERCISE 1] SENSOR DATA PROCESSING
SURNAME = "Soliva"
SEED_NUM = 4

# 1. Generate Sensor Data mock values based on seed
generated_raw_data = [45.2, "ERROR", 12.8, 89.1, "NULL", 5.4, 61.3, 102.5]
valid_readings = []
invalid_count = 0
log = []

log.append("--- EXECUTION LOG STARTED ---")
log.append(f"Initializing sensor array validation for Engineer {SURNAME}.")

# 2. Process and validate data streams
for entry in generated_raw_data:
    try:
        val = float(entry)
        valid_readings.append(val)
        log.append(f"Validated reading: {val}")
    except ValueError:
        invalid_count += 1
        log.append(f"CRITICAL: Suppressed invalid input artifact '{entry}'")

# 3. Classify valid readings according to predefined operating ranges
classification_results = {"Low (0-40)": 0, "Nominal (41-80)": 0, "High (>80)": 0}
for val in valid_readings:
    if val <= 40.0:
        classification_results["Low (0-40)"] += 1
    elif val <= 80.0:
        classification_results["Nominal (41-80)"] += 1
    else:
        classification_results["High (>80)"] += 1

# --- DISPLAY ASSESSMENT DATA ---
print("=== ASSESSMENT DATA ===")
print(f"Generated Sensor Data: {generated_raw_data}")
print(f"Valid/Invalid Results: {len(valid_readings)} Valid | {invalid_count} Invalid")
print(f"Classification Results: {classification_results}")
print("\nExecution Log:")
for entry in log:
    print(f"  {entry}")

print("\nFinal Output:")
print(f"System processing complete. Mean valid payload value: {sum(valid_readings)/len(valid_readings):.2f}°C")
```

```python
# [EXERCISE 2] SIGNAL CHARACTER ANALYSIS
SURNAME = "Soliva"
SEED_NUM = 4

# 1. Generate signal payload string
generated_signal = f"  Signal-Type_A-{SURNAME}#{SEED_NUM}09!!  "

# 2. Process and Normalize
processed_signal = generated_signal.strip().upper()

# 3. Character by Character Parsing
analysis = {"Letters": 0, "Numbers": 0, "Spaces": 0, "Special Chars": 0}
execution_log = []

for char in processed_signal:
    if char.isalpha():
        analysis["Letters"] += 1
        execution_log.append(f"Parsed '{char}': Alpha Token")
    elif char.isdigit():
        analysis["Numbers"] += 1
        execution_log.append(f"Parsed '{char}': Numeric Token")
    elif char.isspace():
        analysis["Spaces"] += 1
        execution_log.append(f"Parsed '{char}': Whitespace Token")
    else:
        analysis["Special Chars"] += 1
        execution_log.append(f"Parsed '{char}': Operator/Special Token")

# --- DISPLAY ASSESSMENT DATA ---
print("=== ASSESSMENT DATA ===")
print(f"Generated Signal: '{generated_signal}'")
print(f"Processed Signal: '{processed_signal}'")
print(f"Character Analysis: {analysis}")
print(f"Signal Classification: Complex Alpha-Numeric Composite Metric")
print("\nExecution Log Sample (First 5 tokens):")
for entry in execution_log[:5]:
    print(f"  {entry}")

print("\nFinal Output:")
print(f"Signal stream decompiled. Total payload lengths verified at {len(processed_signal)} bits.")
```

```python
# [EXERCISE 3] SECURE AUTHENTICATION SYSTEM
SURNAME = "Soliva"
SEED_NUM = 4

# 1. Credentials parameters mapping
generated_password = f"{SURNAME.upper()}{SEED_NUM}" # E.g., SOLIVA4
attempt_limit = SEED_NUM if SEED_NUM >= 3 else 3     # Limit is 4 based on your seed
attempts_made = 0
access_result = "DENIED"
final_system_state = "LOCKED"
execution_log = []

# Mock user entry inputs for simulation testing
simulated_user_inputs = ["HELLO", "SOLIVA", "SOLIVA4"] 

for user_attempt in simulated_user_inputs:
    attempts_made += 1
    execution_log.append(f"Attempt {attempts_made}: User evaluated value '{user_attempt}'")
    
    if user_attempt == generated_password:
        access_result = "GRANTED"
        final_system_state = "UNLOCKED"
        execution_log.append("Success: Core signature authorization verified.")
        break
    else:
        execution_log.append("Warning: Authorization signature mismatch.")
        
    if attempts_made >= attempt_limit:
        execution_log.append("CRITICAL: Exceeded safety attempt parameters. Intruding terminal frozen.")
        break

# --- DISPLAY ASSESSMENT DATA ---
print("=== ASSESSMENT DATA ===")
print(f"Generated Password: {generated_password}")
print(f"Attempt Limit: {attempt_limit}")
print(f"Attempts Made: {attempts_made}")
print(f"Access Result: {access_result}")
print(f"Final System State: {final_system_state}")
print("\nExecution Log:")
for entry in execution_log:
    print(f"  {entry}")

print("\nFinal Output:")
print(f"Security Core Status: Execution terminated with exit status [{final_system_state}]")
```
